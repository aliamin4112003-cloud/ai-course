
# **🔹 Question 2: Simple Calculator**
## Create a Python program that:

### 1- Asks the user to enter two numbers.

### 2-Stores them as variables (make sure to convert them to integers).

### 3-Performs and prints the results of the following operations:

* ## Addition

* ## Subtraction

* ## Multiplication

* ## Division

* ## Power (num1 ** num2)

* ## Modulus (num1 % num2)

* ## Floor division (num1 // num2)






```python
number1 =int(input("pz entter ur number1:"))
number2 =int(input("pz entter ur number2:"))
print(f"Addition:{number1+number2}")
print(f"Subtraction:{number1-number2}")
print(f"Multiplication:{number1*number2}")
print(f"Division:{number1/number2}")
print(f"Power (num1 ** num2):{number1**number2}")
print(f"Modulus (num1 % num2):{number1%number2}")
print(f"Floor division (num1 // num2):{number1//number2}")
```
